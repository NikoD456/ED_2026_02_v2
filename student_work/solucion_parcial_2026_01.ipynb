```python
def Niidea(S):
    n = len(S)
    total = 0
    for j in range(0, n, 2):
        for i in range(0, 2):
            total = total + 2*S[j]*S[i]
    return total

Niidea([1,2,3,4])
```

- 2.1 MinMax

```python
def MinMax(L):
    if len(L) == 1:
        min = L[0]
        max = L[0]
        return min, max

    min, max = MinMax(L[:-1])
    a = L[-1]
    if a < min:
        min = a
    if a > max:
        max = a

    return min, max

MinMax([1,2,3,4,5,6,-9])
```

```python
def picos(L):
    if len(L) < 3:
        return 0
    pico = 1 if L[0] < L[1] > L[2] else 0
    return pico + picos(L[1:]) 

def picos1(L):
    if len(L) < 3:
        return 0
    picos = picos1(L[1:])
    if L[0] < L[1] > L[2]:
        return 1 + picos
    else:
        return picos
    

picos1([0,1,-1,-2,3,0,1,-2])
```

```python
import matplotlib.pyplot as plt

def graficar_lista(datos):
    x = list(range(len(datos)))   # índices: 0, 1, 2, 3, 4...
    y = datos                     # valores: 1, 2, 3, 2, 1...

    plt.plot(x, y, marker='o', linestyle='-', color='b')
    plt.xlabel("Índice")
    plt.ylabel("Valor")
    plt.title("Gráfica de la lista")
    plt.grid(True)
    plt.show()

# Ejemplo
graficar_lista([0,1,-1,-2,3,0,1,-2])
```
